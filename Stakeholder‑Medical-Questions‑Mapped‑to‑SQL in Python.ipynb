##  A Stakeholder Question and an SQL Query : Understanding the difference 

## Overview : 

In data‑driven projects, communication often flows between business stakeholders and technical teams. Although both groups aim to extract value from data, they speak in different “languages.” A stakeholder question expresses a business need in natural language, while an SQL query is the technical instruction used to retrieve the data required to answer that need. Understanding the distinction between the two is essential for accurate requirement gathering, efficient analysis, and effective decision‑making.

Stakeholder Question
A stakeholder question is a business‑oriented inquiry that describes what information is needed, without specifying how to obtain it. It is typically expressed in plain language and focuses on goals, performance, trends, or operational concerns.

In this file, I present multiple examples drawn from medical datasets to clearly demonstrate how a stakeholder’s business question is translated into a MySQL query. Each example guides the reader through the full reasoning process—from interpreting the stakeholder’s intent, to identifying the relevant tables and fields, to writing the SQL needed to extract the correct information. 

This step‑by‑step approach helps beginners understand not only the conceptual difference between a business question and a technical query, but also how to bridge that gap in real‑world data projects. In addition, the file introduces the practical workflow of connecting to a database from Python and executing SQL queries programmatically, giving new data professionals a complete view of how analytical tasks move from business requirements to executable code.

```python

# How to connect MySQL
import mysql.connector
import pandas as pd

conn = mysql.connector.connect(
    host="00.000.000.000",
    user="myownwork",
    password="$=$@B4=B4",
    database="myowndb")
```

```python

# Stakeholder‑Questions 1: 
# I need a breakdown of all male patients with their basic identifying details.

query = "SELECT first_name, last_name, gender FROM patients WHERE gender='M';" # If M is the corresponding column for male

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 2: 
# Give me the roster of patients without any recorded allergies, limited to their first and last names.

query = "SELECT first_name, last_name, allergies FROM patients WHERE allergies is Null;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 3: 
# Identify all patients whose first names start with the letter C

query = "SELECT first_name FROM patients where first_name like 'C%';"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 4: 
# Present the list of patients meeting the 60–80 weight criteria, limited to their first and last names.

query = "SELECT first_name, last_name, weight from patients WHERE weight between 60 and 80 ORDER BY weight ASC;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 5: 
# Present a unified name column showing each patient’s first and last names together.

query = "SELECT CONCAT(first_name, ' ', last_name) AS full_name FROM patients;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 6: 
# Provide a detailed listing of patients with their names and the full province name attached to their profiles.

# Please Note : In a situation where there are two tables namely patients table and provinces table, the following query works fine.

query = "SELECT p.first_name, p.last_name, pr.province_name FROM patients p JOIN province_names pr ON p.province_id= pr.province_id;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 7: 
# Present the total patient count for individuals born in 2010.

query = "SELECT COUNT(*) as total_patients_count FROM patients WHERE YEAR(birth_date) = 2010;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 8: 
# Present the tallest patient on record, showing their first and last names along with their height

query = "SELECT first_name, last_name, height FROM patients WHERE height = (SELECT MAX(height) FROM patients);"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 9: 
# Provide all available patient details for individuals whose patient IDs are 24, 95, 604, 872, and 1031.

query = "SELECT * FROM patients WHERE patient_id IN (24, 95, 604, 872, 1031);"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 10: 
# Present the aggregate number of admissions captured in the system

query = "SELECT COUNT(*) AS aggregate_admissions FROM admissions;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 11: 
# Deliver the complete set of admission details for patients whose admission and discharge occurred on the same day.

# Please Note : The following query works for a table having a column for admission_date and another column for discharge_date

query = "SELECT * FROM admissions WHERE admission_date = discharge_date;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 12: 
# Present the total number of admissions recorded for patient 579.

query = "SELECT COUNT(*) AS total_admissions FROM admissions WHERE patient_id = 579;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 13:
# Present the set of unique cities recorded for patients whose province ID is ‘TH’

query = "SELECT DISTINCT city FROM patients WHERE province_id = 'TH';"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 14:
# Present the names and birth dates of patients whose recorded height and weight are above 140 and 60 respectively.

query = "SELECT first_name, last_name, birth_date, height, weight FROM patients WHERE height > 140  AND weight > 60;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 15:
# Present the set of unique birth years captured in the patients table, sorted from earliest to latest.

query = "SELECT DISTINCT YEAR(birth_date) AS birth_year FROM patients ORDER BY birth_year ASC;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 16:
# Provide the list of unique first names from the patients table that appear only once in the dataset.

query = "SELECT first_name FROM patients GROUP BY first_name HAVING COUNT(*) = 1;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 17:
# Provide the patient ID and first name for all individuals whose first name begins and ends with the letter ‘s’ and contains at least six characters.

query = "SELECT patient_id, first_name FROM patients WHERE first_name LIKE 's%s'AND LENGTH(first_name) >= 6;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 18:
# Present the identifying details (patient ID, first name, last name) for all patients classified under the diagnosis ‘Hypertension’.

# Please Note : The following query works in a situation where there are two tables : patient table and admission table

query = "SELECT p.patient_id, p.first_name, p.last_name FROM patients p JOIN admissions a ON p.patient_id = a.patient_id WHERE a.diagnosis = 'Hypertension';"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 19:
# Provide a list of all patient first names, ordered first by the length of each name and then alphabetically.

query = "SELECT first_name FROM patients ORDER BY LENGTH(first_name), first_name ASC;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 20:
# Provide a consolidated summary showing the count of male patients and female patients, displayed together in one row.

query = "SELECT SUM(gender = 'M') AS total_male, SUM(gender = 'F') AS total_female FROM patients;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 21:
# Produce a single, clear breakdown of gender counts, detailing the total number of male and female patients.

query = "SELECT SUM(gender = 'M') AS total_male, SUM(gender = 'F') AS total_female FROM patients;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 22:
# Provide a report listing each patient_id alongside their diagnosis from the admissions table, 
# and identify all patients who have been admitted multiple times for the same diagnosis.

query = "SELECT patient_id, diagnosis FROM admissions GROUP BY patient_id, diagnosis HAVING COUNT(*) > 1;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 23:
# Provide a summary of patient distribution by city, showing each city alongside its total number of patients. 
# Order the results from the highest to the lowest patient count, and apply an ascending alphabetical sort for cities with identical totals.

query = "SELECT city, COUNT(*) AS total_patients FROM patients GROUP BY city ORDER BY total_patients DESC, city ASC;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 24:
# Provide a consolidated list showing the first name, last name, and role of every individual classified as either a patient or a doctor. 
# The role designation should be explicitly labeled as “Patient” or “Doctor” for clarity.

# Please Note : The following query works where there are two tables : patients table and doctors table.

query = "SELECT first_name, last_name, 'Patient' AS role FROM patients UNION SELECT first_name, last_name, 'Doctor' AS role FROM doctors;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```

```python

# Stakeholder‑Questions 25:
# Provide a ranked list of all recorded allergies, ordered by their overall frequency. 
# Exclude any NULL or unrecorded values to ensure the results reflect only valid allergy entries.

query = "SELECT allergies, COUNT(*) AS total_cases FROM patients WHERE allergies IS NOT NULL GROUP BY allergies ORDER BY total_cases DESC, allergies ASC;"

df = pd.read_sql(query, conn)

# Display results
pd.set_option('display.max_rows', None)   # show all rows
pd.set_option('display.max_columns', None)  # show all columns

df
```
